# 01 - EDA and Baseline Model
Explore the Diabetes 130-US Hospitals dataset and train a first baseline model.

```python
import sys
sys.path.append('../src')
import pandas as pd
import matplotlib.pyplot as plt
from data_prep import load_and_clean

df = load_and_clean('../data/diabetic_data.csv')
df.shape
```

```python
# Class balance - this matters a lot for how we evaluate later
df['readmitted_30d'].value_counts(normalize=True)
```

```python
# Quick look at a few features vs. the target
df.groupby('age')['readmitted_30d'].mean().plot(kind='bar', title='Readmission rate by age bracket')
plt.ylabel('30-day readmission rate')
plt.show()
```

```python
df.groupby('number_inpatient')['readmitted_30d'].mean().plot(title='Readmission rate by prior inpatient visits')
plt.show()
```
